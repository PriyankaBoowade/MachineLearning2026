```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

X = np.array([0.5, 2.3, 2.9])
y = np.array([1.4, 1.9, 3.2])
```

m = Σ(xi - x̄)(yi - ȳ) / Σ(xi - x̄)²
c = ȳ - m·x̄

```python
def ols_linear_regression(X, y):
    """
    Compute linear regression parameters analytically using OLS closed-form formulas.
    """
    # TODO 1.1: Compute mean of X and mean of y
    x_mean = np.mean(X)
    y_mean = np.mean(y)

    # TODO 1.2: Compute numerator and denominator for slope 'm'
    numerator = np.sum((X - x_mean) * (y - y_mean))
    denominator = np.sum((X - x_mean) ** 2)
    m = numerator / denominator

    # TODO 1.3: Compute intercept 'c'
    c = y_mean - m * x_mean

    return m, c
```

```python
def gradient_descent(X, y, alpha=0.05, epochs=1000):
    """
    Optimize linear regression parameters using batch gradient descent.
    """
    m = 0.0
    c = 0.0
    n = len(X)
    loss_history = []

    for epoch in range(epochs):
        # TODO 2.1: Compute linear predictions (y_pred = m * X + c)
        y_pred = m * X + c

        # TODO 2.2: Compute Mean Squared Error (MSE) loss and store in loss_history
        loss = np.mean((y_pred - y) ** 2)
        loss_history.append(loss)

        # TODO 2.3: Compute partial derivatives dm and dc
        dm = (2 / n) * np.sum((y_pred - y) * X)
        dc = (2 / n) * np.sum(y_pred - y)

        # TODO 2.4: Update parameters m and c using gradients and learning rate alpha
        m = m - alpha * dm
        c = c - alpha * dc

    return m, c, loss_history
```

```python
def compute_evaluation_metrics(X, y, y_pred, m):
    """
    Calculate R-squared and statistical significance (p-value) of the slope parameter.
    """
    n = len(X)
    df = n - 2

    # TODO 3.1: Calculate SS_res (residual sum of squares) and SS_tot (total sum of squares)
    ss_res = np.sum((y - y_pred) ** 2)
    ss_tot = np.sum((y - np.mean(y)) ** 2)

    # TODO 3.2: Calculate R-squared (R2)
    r2 = 1 - (ss_res / ss_tot)

    # TODO 3.3: Calculate Standard Error of the slope SE(m)
    residual_variance = ss_res / df
    se_m = np.sqrt(residual_variance / np.sum((X - np.mean(X)) ** 2))

    # TODO 3.4: Compute t-statistic (t = m / se_m)
    t_stat = m / se_m

    # TODO 3.5: Compute two-tailed p-value using stats.t.cdf or stats.t.sf
    p_val = 2 * (1 - stats.t.cdf(abs(t_stat), df))

    return r2, t_stat, p_val
```

```python
# --- Execute OLS ---
m_ols, c_ols = ols_linear_regression(X, y)
y_pred_ols = m_ols * X + c_ols
r2_ols, t_ols, p_ols = compute_evaluation_metrics(X, y, y_pred_ols, m_ols)

print("===================== OLS RESULTS =====================")
print(f"Equation   : y = {m_ols:.4f}x + {c_ols:.4f}")
print(f"R-squared  : {r2_ols:.4f}")
print(f"t-statistic: {t_ols:.4f}")
print(f"p-value    : {p_ols:.4f}\n")

# --- Execute Gradient Descent ---
alpha = 0.05
epochs = 1500
m_gd, c_gd, losses = gradient_descent(X, y, alpha=alpha, epochs=epochs)
y_pred_gd = m_gd * X + c_gd
r2_gd, t_gd, p_gd = compute_evaluation_metrics(X, y, y_pred_gd, m_gd)

print("================ GRADIENT DESCENT RESULTS ================")
print(f"Equation   : y = {m_gd:.4f}x + {c_gd:.4f}")
print(f"R-squared  : {r2_gd:.4f}")
print(f"t-statistic: {t_gd:.4f}")
print(f"p-value    : {p_gd:.4f}\n")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Subplot 1: Regression Fit Comparison
# TODO 5.1: Plot actual data points as scatter plot
axes[0].scatter(X, y, color="black", label="Actual Data", zorder=5)

# TODO 5.2: Generate x_vals using np.linspace(0, 3.5, 100) and plot OLS & GD lines
x_vals = np.linspace(0, 3.5, 100)
axes[0].plot(x_vals, m_ols * x_vals + c_ols, color="blue", label="OLS Fit")
axes[0].plot(x_vals, m_gd * x_vals + c_gd, color="red", linestyle="--", label="Gradient Descent Fit")

axes[0].set_title("OLS vs Gradient Descent Regression Line")
axes[0].set_xlabel("Height (X)")
axes[0].set_ylabel("Weight (y)")
axes[0].legend()
axes[0].grid(True)

# Subplot 2: Loss Convergence Curve
# TODO 5.3: Plot epoch index vs. loss_history
axes[1].plot(range(epochs), losses, color="green")

axes[1].set_title("MSE Loss Convergence over Epochs")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Mean Squared Error (Loss)")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

## Lab Report & Viva Questions

**1. Analytical vs. Iterative Comparison:**
Do the values of (m, c) and R² from Gradient Descent match those from OLS? If there's a slight difference, why?

Yeah they're basically almost the same, just tiny tiny differences after the decimal point. That's cuz Gradient Descent isn't solving it exactly like OLS does — its more like it keeps taking small steps to get closer and closer to the right answer, so after enough epochs it gets really close but not 100% exact like the direct formula.

**2. Interpreting R² vs. p-value for Small Sample Sizes (n=3):**
Notice the R² score is high (>0.80), yet the p-value may exceed 0.05. Explain why.

With only 3 points, theres barely any data to be sure the pattern is "real" and not just random luck. R² just tells you how well the line fits THESE 3 points, but the p-value is checking if we can be statistically confident the relationship isnt a coincidence — and with df=1 (basically no wiggle room), its really hard to prove that confidently even if the fit looks good visually.

**3. Hypothesis Testing:**
State the null hypothesis (H0) and alternative hypothesis (H1) tested by the t-statistic on the slope m.

H0 is that m = 0, meaning there's no real relationship between height and weight. H1 is that m ≠ 0, meaning there IS a real relationship. The t-statistic and p-value basically help us decide whether to reject H0 or not.

**4. Learning Rate Sensitivity:**
What happens if you increase alpha to 1.0? What if you decrease it to 0.0001?

If alpha is too big (like 1.0), the steps are too large and the model can overshoot the best answer, or even blow up and never settle down (the loss might bounce around or explode instead of going down smoothly). If alpha is too small (like 0.0001), the steps are tiny, so it takes WAY more epochs to actually get close to the right answer — it'll converge but super slowly.

**5. Computational Scalability:**
Why is Gradient Descent preferred over OLS for datasets with millions of features/samples?

OLS needs to compute (X^T X)^-1, which is a matrix inversion — and that gets really expensive (slow and memory-heavy) as the number of features grows, since the size of that matrix grows too. Gradient Descent doesn't need to invert anything, it just takes small steps repeatedly, so it scales way better to huge datasets even if it takes more iterations to get there.
